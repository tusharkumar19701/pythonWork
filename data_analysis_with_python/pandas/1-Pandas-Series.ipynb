# Pandas Series

```python
import pandas as pd
import numpy as np
```

We'll start analyzing "The Group of Seven". Which is a political formed by Canada, France, Germany, Italy, Japan, the United Kingdom and the United States. We'll start by analyzing population, and for that, we'll use a pandas.Series object.

```python
# in millions
g7_pop = pd.Series([35.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523])
```

```python
g7_pop
```

Someone might not know we're representing population in millions of inhabitants. Series can have a name, to better document the purpose of the Series:

```python
g7_pop.name = 'G7 Population in millions'
```

```python
g7_pop
```

```python
g7_pop.dtype
```

```python
g7_pop.values
```

They're actually backed by numpy arrays:

```python
type(g7_pop.values)
```

And they look like simple Python lists or Numpy Arrays. But they're actually more similar to Python dicts.

A Series has an index, that's similar to the automatic index assigned to Python's lists:

```python
g7_pop
```

```python
g7_pop[0]
```

```python
g7_pop.index
```

```python
l = ['a','b','c']
```

```python
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States',
]
```

```python
g7_pop
```

We can say that Series look like "ordered dictionaries". We can actually create Series out of dictionaries:

```python
pd.Series({
    'Canada': 35.467,
    'France': 63.951,
    'Germany': 80.94,
    'Italy': 60.665,
    'Japan': 127.061,
    'United Kingdom': 64.511,
    'United States': 318.523
},name='G7 Population in millions')
```

```python
pd.Series([35.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523],
         index=['Canada', 'France', 'Germany', 'Italy', 'Japan', 'United Kingdom',
       'United States'],
         name='G7 Population in millions')
```

You can also create Series out of other series, specifying indexes:

```python
pd.Series(g7_pop, index=['France','Germany','Italy'])
```

## Indexing

Indexing works similarly to lists and dictionaries, you use the index of the element you're looking for:

```python
g7_pop
```

```python
g7_pop['Canada']
```

```python
g7_pop.iloc[0]
```

```python
g7_pop.iloc[-1]
```

Selecting multiple elements at once:

```python
g7_pop[['France','Italy']]
```

```python
g7_pop.iloc[[0,1]]
```

(The result is another Series)

Slicing also works, but important, in Pandas, the upper limit is also included:

```python
g7_pop['Canada':'Italy']
```

## Conditional selection (boolean arrays)

The same boolean array techniques we saw applied to numpy arrays can be used for Pandas Series:

```python
g7_pop
```

```python
g7_pop > 70
```

```python
g7_pop[g7_pop > 70]
```

```python
g7_pop[g7_pop > g7_pop.mean()]
```

```python
g7_pop.std()
```

## Operations and methods

Series also support vectorized operations and aggregation functions as Numpy:

```python
g7_pop
```

```python
g7_pop * 1000000
```

```python
g7_pop.mean()
```

```python
np.log(g7_pop)
```

```python
g7_pop['France':'Italy'].mean()
```

## Boolean arrays

(Work in the same way as numpy)

```python
g7_pop
```

```python
g7_pop > 80
```

```python
g7_pop[(g7_pop > 80) & (g7_pop < 200)]
```

## Modifying series

```python
g7_pop['Canada'] = 40.5
```

```python
g7_pop
```

```python
g7_pop.iloc[-1] = 500
```

```python
g7_pop[g7_pop < 70]
```
